# Benchmark bounded clique enumeration

Compare NetworkX's bounded all-clique enumeration with a bitset adjacency-intersection enumerator on cached protein contact graphs. The benchmark checks that both methods return exactly the same cliques, then reports median enumeration time for each graph. It times clique enumeration only; representative-pair traversal and feature pairing are not included.

```python
from pathlib import Path
import sys

PROJECT_ROOT = next(
    (path for path in (Path.cwd().resolve(), *Path.cwd().resolve().parents)
     if (path / "notebooks" / "src" / "enzyme_workflow").is_dir()),
    None,
)
if PROJECT_ROOT is None:
    raise RuntimeError("Could not find the repository root from this notebook's working directory")
sys.path.insert(0, str(PROJECT_ROOT / "notebooks" / "src"))

from enzyme_workflow.common import add_project_sources
PROJECT_ROOT = add_project_sources(PROJECT_ROOT)
```

```python
# Experiment settings
SAMPLE_COUNT = 5
MAX_SIMPLEX_DIMENSION = 3
REPEATS = 5

from enzyme_workflow.clique_benchmark import benchmark_clique_enumeration
results, summary = benchmark_clique_enumeration(
    PROJECT_ROOT,
    sample_count=SAMPLE_COUNT,
    max_simplex_dimension=MAX_SIMPLEX_DIMENSION,
    repeats=REPEATS,
)
display(summary)
display(results)
```
