# AbstractGraph-style feature hashing benchmark

Compare three paths: current BLAKE2b with standard JSON, current BLAKE2b with `orjson`, and an experiment using AbstractGraph's `hash_value` (SHA-256 over canonical type-tagged JSON). All paths enumerate the same representative-paired simplex occurrences. The notebook verifies that the two current BLAKE2b paths produce identical coordinates and that the AbstractGraph path preserves the total occurrence count; its coordinate values differ because the hash function and encoding differ.

This is an experiment only. It requires `orjson` and uses the local AbstractGraph source checkout (or an installed `abstractgraph` package); it does not change the production vectorizer.

```python
from pathlib import Path
import sys

PROJECT_ROOT = next(
    (path for path in (Path.cwd().resolve(), *Path.cwd().resolve().parents)
     if (path / 'notebooks' / 'src' / 'enzyme_workflow').is_dir()),
    None,
)
if PROJECT_ROOT is None:
    raise RuntimeError("Could not find the repository root from this notebook's working directory")
sys.path.insert(0, str(PROJECT_ROOT / 'notebooks' / 'src'))

from enzyme_workflow.common import add_project_sources
PROJECT_ROOT = add_project_sources(PROJECT_ROOT)
from enzyme_workflow.abstractgraph_hash_benchmark import benchmark_abstractgraph_hashing
```

```python
SAMPLE_COUNT = 3
MAX_SIMPLEX_DIMENSION = 3
MAX_REPRESENTATIVE_DISTANCE = 1
NBITS = 14
REPEATS = 3
```

```python
results, summary = benchmark_abstractgraph_hashing(
    PROJECT_ROOT,
    sample_count=SAMPLE_COUNT,
    max_simplex_dimension=MAX_SIMPLEX_DIMENSION,
    max_distance=MAX_REPRESENTATIVE_DISTANCE,
    nbits=NBITS,
    repeats=REPEATS,
)
display(summary)
display(results)
```
